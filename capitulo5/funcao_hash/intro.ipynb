## **Introdução - Tabela Hash**

### **Função Hash**

Uma `função hash`é um algoritmo matemático que recebe uma string (uma sequência de bytes) e retorna um número correspondente.

Para ser considerada uma boa função hash, ela precisa seguir duas regras:

- Ela deve ser consistente, ou seja, deve retornar o mesmo número sempre que for passado a mesma string. Se ela retornar um número diferente, a tabela hash não funcionará corretamente.

- A função deve mapear diferentes palavras para diferentes números. No melhor caso, cada palavra diferente deveria ser mapeada e ligada a um número diferente, evitando colisões.

### **Tabela Hash**

Uma função hash conectada a um array comum na memória é uma estrutura de dados chamada de `tabela hash`.

As `tabelas hash` utilizam uma função hash para indicar, de maneira inteligente, onde armazear os elementos. 

**Exemplo**: 

- O programa passa uma string qualquer pela função hash.

- A função retorna um número, por exemplo o número 3.

- O programa vai direto no índice 3 do array e armazena um determinado valor.



- Para realizar a busca, é passado novamente a string para a função hash, que retorna o mesmo valor anterior (Ex: 3).

- O computador olha direto no índice 3 do array e retorna o valor armazenado.

**Por isso que buscas em tabela hash custam `O(1)`.**

### **Tabelas Hash em Python**

No Python, as tabelas hash são implementadas nativamente sob o nome de **Dicionários (`dict`)**.

Uma tabela hash mapeia chaves e valores.

```python
# Exemplo - Tabela Hash (Dicionários)

tabela_precos = {}

tabela_precos['maçã'] = 4.5
tabela_precos['leite'] = 10.0
tabela_precos['uva'] = 5.75

print(tabela_precos)
```
