## **Colisões**

### **1. O que é uma colisão?**

Uma colisão acontece quando a sua função hash aponta dois itens diferentes para o mesmo índice da tabela, ou seja, duas chaves são indicadas para o mesmo espaço de memória.

Uma maneira simples de resolver é a **Lista Encadeada**: se diversas chaves mapeiam para o mesmo índice, inicie uma lista encadeada neste espaço.

Se a lista encadeada for pequena, não haverá nenhum problema. Mas, para isso, é necessário uma **boa função hash**.

### **2. Fator de Carga**

O Fator de Carga é a métrica que nos diz se a tabela hash precisa de **redimensionamento ou não**, ou seja, mede a proporção de espaços vazios.



Um fator de carga maior do que 1 indica que existe mais itens do que espaços no array, acarretando em uma **colisão**.

Uma boa regra geral é: redimensionar o array quando o fator de carga for maior do que 0,7.

### **3. Desempenho**

Quando o fator de carga fica em torno de 0.7, é necessário uma operação de **Redimensionamento**.

- É criada uma nova tabela hash, normalmente com o drobro do tamanho da original.

- Cada item da tabela anterior é reinserido na nova tabela, recalculando o hash de cada um.

Se o dimensionamento não ocorrer, o fator de carga pode ultrapassar o valor de 1, as listas encadeadas vão ficar enormes e a busca, que deveria ser O(1) vai degradar para O(n).

```python
# Exemplo de colisão

tabela_hash = [None for _ in range(10)]

def funcao_hash(string):

    idx = len(string)

    if idx >= len(tabela_hash):

        print("String muito longa.")

        return

    if tabela_hash[idx] != None:

        print(f"Alerta de colisão, índice {idx} já está ocupado!")
        tabela_hash[idx] = string
        print("Ocorreu uma sobrescrita!")

    else:

        tabela_hash[idx] = string
        print(f"String {string} inserida no índice {idx}")

funcao_hash("Elis")       
funcao_hash("Guthierry")   

print("\nEstado da Tabela Hash:")
print(tabela_hash)
print()

funcao_hash("Elen")

print(tabela_hash)
print()
```

Uma boa função hash distribui os valores no array simetricamente, ou seja, ela não agrupa valores em um mesmo local.
