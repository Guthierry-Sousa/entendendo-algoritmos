```python
import numpy as np
import matplotlib.pyplot as plt
import math as mt
```

```python
# Valores de n -> Número de Dados
n = [i for i in range(1, 30)]
```

```python
# Tempos

O_log_n = [np.log2(i) for i in n]
O_n = n
O_n_log_n = [i*np.log2(i) for i in n]
O_n_2 = [i**2 for i in n]
O_n_fat = [mt.factorial(i) for i in n]
```

```python
plt.figure(figsize=(10,5))

plt.title("Tempo O(log n)")

plt.plot(n, O_log_n, color = 'black')
plt.xlabel("n - número de dados")
plt.ylabel("número de operações (Big O)")

plt.ylim((0, 200))

plt.grid(alpha = 0.3)

plt.show()
```

```python
plt.figure(figsize=(10,5))

plt.title("Tempo O(n)")

plt.plot(n, O_n, color = 'black')
plt.xlabel("n - número de dados")
plt.ylabel("número de operações (Big O)")

plt.ylim((0, 200))

plt.grid(alpha = 0.3)

plt.show()
```

```python
plt.figure(figsize=(10,5))

plt.title("Tempo O(n * log n)")

plt.plot(n, O_n_log_n, color = 'black')
plt.xlabel("n - número de dados")
plt.ylabel("número de operações (Big O)")

plt.ylim((0, 200))

plt.grid(alpha = 0.3)

plt.show()
```

```python
plt.figure(figsize=(10,5))

plt.title("Tempo O(n²)")

plt.plot(n, O_n_2, color = 'black')
plt.xlabel("n - número de dados")
plt.ylabel("número de operações (Big O)")

plt.ylim((0, 200))

plt.grid(alpha = 0.3)

plt.show()
```

```python
plt.figure(figsize=(10,5))

plt.title("Tempo O(n!)")

plt.plot(n, O_n_fat, color = 'black')
plt.xlabel("n - número de dados")
plt.ylabel("número de operações (Big O)")

plt.ylim((0, 200))

plt.grid(alpha = 0.3)

plt.show()
```
