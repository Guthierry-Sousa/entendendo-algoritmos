### **Algoritmos e Pesquisa Binária**

#### **Algoritmos**

Um `algoritmo` é um conjunto de instruções que descrevem como realizar uma tarefa ou resolver um problema.



#### **Pesquisa Binária**

`Pesquisa Binária` é um algoritmo. Sua entrada é uma lista ordenada de elementos.

Se o elemento que você está buscando está na lista, a pesquisa binária retorna a sua localização. Caso contrário, a pesquisa binária retorna `None`.

A `Pesquisa Simples` é quando a busca ocorre sequencialmente, do início ao fim da lista. É uma busca linear.

De maneira geral para uma lista de **n** números, a pesquisa binária precisa de, no máximo, $log(n)$ tentativas para retornar o valor correto, enquanto a pesquisa simples precisa de $n$ etapas.
