## **Algoritmo de Dijkstra**

O algoritmo de Dijkstra resolve o problema do **caminho mais curto a partir de uma única origem em grafos ponderados (com peso)**.

Para calcular o caminho mínimo em um grafo ponderado é utilizado o **algoritmo de Dijkstra**. Já para calcular o caminho mínimo em um grafo não ponderado, é utilizado a **pesquisa em largura (BFS)**.

O algoritmo de Dijkstra faz a suposição que uma vez que um vértice é processado, isso significa que não há uma maneira mais barata de chegar até ele. Essa suposição só funciona caso não haja arestas com pesos negativos. Portanto, **o algoritmo de Dijkstra não funciona com pesos negativos**.

```python
# Grafo

grafo = {}

grafo['A'] = {}
grafo['A']['B'] = 2
grafo['A']['C'] = 5

grafo['B'] = {}
grafo['B']['C'] = 8
grafo['B']['D'] = 7

grafo['C'] = {}
grafo['C']['D'] = 2
grafo['C']['E'] = 4

grafo['D'] = {}
grafo['D']['F'] = 1

grafo['E'] = {}
grafo['E']['D'] = 6
grafo['E']['F'] = 3

grafo['F'] = {}

for k, v in grafo.items():

    print(f'{k}:{v}')
```

```python
# Tabela com os pesos 

pesos = {}

pesos['A'] = 0
pesos['B'] = float('inf')
pesos['C'] = float('inf')
pesos['D'] = float('inf')
pesos['E'] = float('inf')
pesos['F'] = float('inf')

for k, v in pesos.items():

    print(f'{k}:{v}')
```

```python
def encontrar_menor_peso(pesos: dict, visitados: set) -> str:

    menor_peso = float('inf')
    no = None

    for n, p in pesos.items():

        if (p < menor_peso) and (n not in visitados):

            menor_peso = p
            no = n

    return no

def algoritmo_dijkstra(grafo: dict[dict], pesos: dict, no_final: str) -> list:

    visitados = set()
    pais = {}

    vertice = encontrar_menor_peso(pesos, visitados)

    while vertice:

        if vertice == no_final:

            caminho = []
            atual = no_final

            while atual:

                caminho.append(atual)
                atual = pais.get(atual)

            return caminho[::-1]

        vizinhos = grafo.get(vertice)
        peso = pesos.get(vertice)

        for v, p in vizinhos.items():

            novo_peso = p + peso

            if novo_peso < pesos.get(v):

                pesos[v] = novo_peso
                pais[v] = vertice

        visitados.add(vertice)
        vertice = encontrar_menor_peso(pesos, visitados)

    return None
```

```python
caminho_minimo = algoritmo_dijkstra(grafo, pesos, 'F')
custo_minimo = pesos['F']

print(f'Caminho mais curto: {caminho_minimo}')
print(f'Custo total: {custo_minimo}')
```

### **Recapitulando**:

- A pesquisa em largura é usada para calcular o caminho mínimo para um grafo não ponderado.

- O algoritmo de Dijkstra é usado para calcular caminho mínimo para um grafo ponderado.

- O algoritmo de Dijkstra funciona quando todos os pesos são positivos.

- Se o grafo tiver pesos negativos, use o algoritmo de Bellman-Ford.
