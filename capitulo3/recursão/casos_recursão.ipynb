# **Caso-base e caso recursivo**

### **Recursão**: é quando uma função chama a si mesma.

Devido ao fato de a função recursiva chamar a si mesma, é mais fácil escreve-lâ erroneamente e acabar em um loop infinito.

#### **Exemplo**

```python
# Essa função ficará executando para sempre: loop infinito!
def contagem_regressiva(num: int):

    print(num)

    contagem_regressiva(num - 1)

contagem_regressiva(10)
```

Quando você escreve uma função recursiva, deve informar quando a recursão deve parar. É por isso que toda *toda função recursiva tem duas partes: o caso-base e o caso recursivo*.

### **Caso recurivo:** é quando a função chama a si mesma.

### **Caso-base:** é quando a função não chama a si mesma novamente, de forma que o programa não se torna um loop infinito.

```python
# Adicionando o caso-base
def contagem_regressiva(num: int):

    if num < 1: # Caso-base

        return
    
    print(num, end=" ")

    contagem_regressiva(num - 1) # Caso recursivo

contagem_regressiva(10)
```

```python
contagem_regressiva(100)
```

### **Recapitulando**

- Resursão é quando uma função chama a si mesma;
- Funções recursivas possuem caso-base e caso recursivo;
- Um pilha tem duas operações: pop e push;
- Todas as chamadas de função vão para a pilha de chamada;
- A pilha pode ficar extremamente grande e ocupar memória que não devia (stack overflow).
