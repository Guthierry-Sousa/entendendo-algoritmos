# **Pilha de Chamada (call stack)**

### **Pilha**

A *pilha* é uma estrutura de dados que permite apenas duas operações: *push (inserir)* e *pop (remover)*.

A *pilha* (stack) possui um comportamento chamado de **LIFO (*last in, first out*)**.

- **O último a entrar é o primeiro a sair**

```python
pilha = []

# Push - Inserção (append)
pilha.append("Entendendo Algoritmos")
pilha.append("Código Limpo")
pilha.append("O Programador Pragmático")

print(f"Pilha Atual: {pilha}")

# Pop - Remoção (pop)
livro_removido = pilha.pop()
print(f"Livro Removido: {livro_removido}")

livro_removido = pilha.pop()
print(f"Livro Removido: {livro_removido}")

print(f"Pilha Atual: {pilha}")
```

### **Pilha de Chamada**

A **Pilha de Chamadas (Call Stack)** é o mecanismo que as linguagens de programação usam para organizar a execução das funções de um programa.

Toda vez que uma função é chamada no código, o "computador" faz o seguinte:

1. Pausa o que quer que estivesse fazendo na linha atual.
2. Cria um bloco de memória chamado *Frame de Função* (que guarda as variáveis locais daquela função e a linha onde ela parou).
3. Empilha esse bloco no topo da pilha (PUSH).
4. Inicia a execução da função.

Quando a função chega ao fim, o "computador" tira esse bloco do topo da pilha (POP) e volta a executar o código anterior de onde parou.

#### **Exemplo do Livro**:

1. A função `sauda("Maggie")` é chamada e seu **frame de ativação** é empilhado na pilha.
2. `sauda` imprime **"Hello Maggie!"**.
3. A função `sauda2("Maggie")` é chamada e um novo **frame de ativação** é empilhado sobre o de `sauda`.
4. `sauda2` imprime **"Como vai Maggie?"**.
5. `sauda2` termina sua execução, seu frame é removido da pilha e o controle retorna para `sauda`.
6. `sauda` imprime **"Preparando a despedida"**.
7. A função `tchau()` é chamada e seu **frame de ativação** é empilhado sobre o de `sauda`.
8. `tchau` imprime **"Ok, bye bye..."**.
9. `tchau` termina sua execução, seu frame é removido da pilha e o controle retorna para `sauda`.
10. `sauda` termina sua execução, seu frame é removido da pilha e a pilha de chamadas fica vazia.


```python
def sauda(nome: str):
    print(f"Hello {nome}!\n")

    sauda2(nome)

    print(f"Preparando a despedida\n")

    tchau()
    
def sauda2(nome: str):
    print(f"Como vai {nome}? \n")

def tchau():
    print(f"Ok, bye bye...")

sauda("Maggie")
```
