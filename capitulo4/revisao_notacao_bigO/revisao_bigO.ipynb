### **Revisando a Notação Big O**

### **Tempos de Execução mais comuns**

- Pesquisa Binária -> O(log_n)

- Pesquisa Simples -> O(n)

- Quicksort -> O(n * log_n)

- Selection sort -> O(n²)

- Caixeiro Viajante -> O(n!)

### **Merge Sort vs. Quicksort**

Quando escrevemos a complexidade de um algoritmo usando a notação Big O, como **O(n)**, estamos, na prática, ignorando fatores constantes. Ou seja, uma complexidade como

$$
O(c \cdot n)
$$

é representada simplesmente como

$$
O(n),
$$

em que (c) é uma constante positiva que representa um fator fixo no tempo de execução.

Na maioria dos casos, essa constante não influencia significativamente o crescimento do algoritmo para entradas muito grandes. Por isso, ela é omitida na notação Big O.

Entretanto, em algumas situações, a diferença entre essas constantes pode ter impacto no desempenho prático. Um exemplo clássico é a comparação entre **Quicksort** e **Merge Sort**.

Ambos possuem complexidade assintótica de

$$
O(n \log n),
$$

mas seus tempos de execução podem ser representados de forma mais detalhada como

$$
\text{Quicksort} = O(c_1 \cdot n \log n)
$$

$$
\text{Merge Sort} = O(c_2 \cdot n \log n),
$$

em que

$$
c_1 < c_2.
$$

Isso significa que, embora ambos pertençam à mesma classe de complexidade assintótica, o Quicksort geralmente executa menos operações para uma mesma entrada, tornando-o mais rápido na prática.

Além disso, o Quicksort apresenta excelente desempenho no **caso médio**, que é o cenário encontrado com maior frequência em aplicações reais. Embora seu pior caso seja

$$
O(n^2),
$$

esse comportamento é raro quando o pivô é escolhido de maneira adequada (por exemplo, de forma aleatória ou utilizando a mediana de três elementos). Por esse motivo, o Quicksort costuma superar o Merge Sort em velocidade na maioria das situações.



```python
import random as rd
```

```python
# Quicksort

def quicksort(lista: list) -> list:

    if len(lista) < 2:

        return lista
    
    idx = rd.randint(0, len(lista) - 1)
    pivo = lista[idx]

    lista_aux = lista[:idx] + lista[idx+1:]
    arr_lower = [i for i in lista_aux if i <= pivo]
    arr_upper = [i for i in lista_aux if i > pivo]

    return quicksort(arr_lower) + [pivo] + quicksort(arr_upper)

lista = [33, 10, 15, 7, 33, 33, 55, 9, -10, 22, 10, -2, 89]

print(quicksort(lista))
```

### **Recapitulando**

- A estratégia DC funciona por meio da divisão do problema em problemas menores. Se você estiver utilizando DC em uma lista, o caso-base provavelmente será um array vazio ou um array com apenas um elemento.

- Se você estiver implementando o quicksort, escolha um elemento aleatório como o pivô. O tempo de execução médio do quicksort é O(n log n)!

- A constante. na notação Big O, pode ser relevante em alguns casos. Esta é arazaão pela qual o quicksort é mais rápido do que merge sort.

- A constante dificilmente será relevante na comparação entre pesquisa simples e pesquisa binária, pois O(log n) é muito mais rápido do que O(n) quando sua lista tende ao infinito de tamanho.
