## **Notação Big O - Anotações Complementares**

A notação Big O descreve como o tempo de execução de um algoritmo cresce conforme o tamanho da entrada aumenta. Ela não mdede o tempo exato em segundos, mas sim a taxa de crescimento.

Essa análise é importante, pois dois algoritmos podem resolver o mesmo problema, mas um pode ficar impraticavelmente lento com muitos dados enquanto o outro continua rápido.

### **Principais Tempos de Execução**

#### **O(1) - Tempo constante**

```python
def primeiro_elemento(lista: list):
    return lista[0]

print(primeiro_elemento(["a", "b", "c"]))
```

#### **O(log n) - Logarítmica**

```python
def busca_binaria(lista: list[int], item: int) -> int:

    menor, maior = 0, len(lista) - 1

    while(menor <= maior):

        meio = (menor + maior) // 2

        if lista[meio] == item:

            return meio

        elif item > lista[meio]:

            menor = meio + 1

        else:

            maior = meio - 1

    return None

print(busca_binaria([1, 2, 3, 4, 5], 4))
```

#### **O(n) - Linear**

```python
def busca_simples(lista, item) -> int:

    for i in range(len(lista)):

        if lista[i] == item:

            return i
    
    return None

print(busca_simples([10, 3, 5, 1, 3, 10, -9, 36, 8, 3, 10], 4))
print(busca_simples([10, 3, 5, 1, 3, 10, -9, 36, 8, 3, 10], 3))
```

#### **O(n log n) - Log Linear**

```python
import random as rd

def quicksort(lista: list) -> list:

    if len(lista) < 2:

        return lista
    
    idx = rd.randint(0, len(lista) - 1)
    num = lista[idx]

    nova_lista = lista[:idx] + lista[idx + 1:]

    lower = [i for i in nova_lista if i <= num]
    upper = [i for i in nova_lista if i > num]

    return quicksort(lower) + [num] + quicksort(upper)

print(quicksort([10, 3, 5, 1, 3, 10, -9, 36, 8, 3, 10]))
```

#### **O(n²) - Quadrática**

```python
def tem_duplicado(lista: list):

    for i in range(len(lista)):

        for j in range(1 + i, len(lista)):

            if lista[i] == lista[j]:

                return True
            
    return False

print(tem_duplicado([10, 3, 5, 1, 3, 10, -9, 36, 8, 3, 10]))
```
